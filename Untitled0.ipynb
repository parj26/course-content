```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.inspection import permutation_importance

# 1. Identify all potential brain volume features (excluding IDs and targets)
exclude_cols = ['dbGaP_Subject_ID', 'SUBJID', 'Race', 'Sex', 'age_at_cnb',
                'LNB_TP', 'LNB_TP2', 'LNB_FP2', 'LNB_RTC2', 'age_at_cnb_squared', 'age_sex_interaction']
brain_features = [col for col in df_processed.columns if col not in exclude_cols]

# 2. Update feature set to include ALL available brain volumes + engineered terms
FULL_FEATURE_SET = brain_features + ['age_at_cnb', 'Sex', 'age_at_cnb_squared', 'age_sex_interaction']

X_full = df_processed[FULL_FEATURE_SET]
y_full = df_processed[TARGET_COLUMN_MODEL]

# Split data
X_train_full, X_test_full, y_train_full, y_test_full = train_test_split(
    X_full, y_full, test_size=0.2, random_state=42
)

# 3. Train a powerful non-linear model (Gradient Boosting)
gb_model = HistGradientBoostingRegressor(random_state=42)
gb_model.fit(X_train_full, y_train_full)

# Evaluate
y_pred_gb = gb_model.predict(X_test_full)
r2_gb = r2_score(y_test_full, y_pred_gb)
mae_gb = mean_absolute_error(y_test_full, y_pred_gb)

print(f"--- Advanced Gradient Boosting Model (All Features) ---")
print(f"R-squared: {r2_gb:.4f}")
print(f"MAE: {mae_gb:.4f}")
print(f"Number of features used: {len(FULL_FEATURE_SET)}")
print(f"\nImprovement over previous Ridge: {r2_gb - r2_inter:.4f}")
```

```python
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler

# 1. Scaling is essential for SVR
scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_train_scaled = scaler_X.fit_transform(X_train_full)
X_test_scaled = scaler_X.transform(X_test_full)

# We scale the target for the model but will inverse it for evaluation
y_train_scaled = scaler_y.fit_transform(y_train_full.values.reshape(-1, 1)).flatten()

# 2. Train a non-linear SVR (RBF kernel)
svr_model = SVR(kernel='rbf', C=1.0, epsilon=0.1)
svr_model.fit(X_train_scaled, y_train_scaled)

# 3. Predict and transform back to original scale
y_pred_scaled = svr_model.predict(X_test_scaled)
y_pred_svr = scaler_y.inverse_transform(y_pred_scaled.reshape(-1, 1)).flatten()

# Evaluate
r2_svr = r2_score(y_test_full, y_pred_svr)
mae_svr = mean_absolute_error(y_test_full, y_pred_svr)

print(f"--- Support Vector Regression (SVR) Results ---")
print(f"R-squared: {r2_svr:.4f}")
print(f"MAE: {mae_svr:.4f}")
print(f"\nDifference from previous best (Ridge): {r2_svr - r2_inter:.4f}")
```

```python
from sklearn.model_selection import GridSearchCV

# Define a parameter grid for SVR tuning
param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': ['scale', 'auto', 0.01, 0.1],
    'epsilon': [0.01, 0.1, 0.2]
}

# Initialize GridSearchCV
grid_svr = GridSearchCV(SVR(kernel='rbf'), param_grid, cv=5, scoring='r2', n_jobs=-1)
grid_svr.fit(X_train_scaled, y_train_scaled)

# Best model
best_svr = grid_svr.best_estimator_

# Evaluate the optimized model
y_pred_opt_scaled = best_svr.predict(X_test_scaled)
y_pred_opt = scaler_y.inverse_transform(y_pred_opt_scaled.reshape(-1, 1)).flatten()

r2_opt = r2_score(y_test_full, y_pred_opt)
mae_opt = mean_absolute_error(y_test_full, y_pred_opt)

print(f"--- Optimized SVR Results ---")
print(f"Best Parameters: {grid_svr.best_params_}")
print(f"R-squared: {r2_opt:.4f}")
print(f"MAE: {mae_opt:.4f}")
print(f"\nImprovement over default SVR: {r2_opt - r2_svr:.4f}")
```

```python
from sklearn.inspection import permutation_importance

# Calculate permutation importance on the test set
result = permutation_importance(
    best_svr, X_test_scaled, y_test_full, n_repeats=10, random_state=42, n_jobs=-1
)

# Organize and display the results
sorted_idx = result.importances_mean.argsort()[::-1]

print("--- Feature Importance (Permutation) ---")
for i in sorted_idx:
    print(f"{FULL_FEATURE_SET[i]:<35} : {result.importances_mean[i]:.4f} +/- {result.importances_std[i]:.4f}")

# Plotting
plt.figure(figsize=(10, 6))
plt.barh([FULL_FEATURE_SET[i] for i in sorted_idx[:10]], [result.importances_mean[i] for i in sorted_idx[:10]])
plt.xlabel("Permutation Importance (R-squared decrease)")
plt.title("Top 10 Most Influential Features for Memory Prediction")
plt.gca().invert_yaxis()
plt.show()
```

```python
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import RidgeCV

# Define the base models for stacking
estimators = [
    ('svr', SVR(C=1, epsilon=0.2, gamma=0.01)),
    ('ridge', Ridge(random_state=42)),
    ('rf', RandomForestRegressor(n_estimators=100, random_state=42))
]

# Initialize the Stacking Regressor with a Ridge meta-model
stack_model = StackingRegressor(
    estimators=estimators,
    final_estimator=RidgeCV(),
    cv=5,
    n_jobs=-1
)

# Train the stacked model on the full feature set
stack_model.fit(X_train_scaled, y_train_scaled)

# Evaluate
y_pred_stack_scaled = stack_model.predict(X_test_scaled)
y_pred_stack = scaler_y.inverse_transform(y_pred_stack_scaled.reshape(-1, 1)).flatten()

r2_stack = r2_score(y_test_full, y_pred_stack)
mae_stack = mean_absolute_error(y_test_full, y_pred_stack)

print(f"--- Ensemble Stacking Model Results ---")
print(f"R-squared: {r2_stack:.4f}")
print(f"MAE: {mae_stack:.4f}")
print(f"\nImprovement over optimized SVR: {r2_stack - r2_opt:.4f}")
```

```python
from sklearn.neural_network import MLPRegressor

# 1. Prepare data using the full feature set
scaler_X_mlp = StandardScaler()
scaler_y_mlp = StandardScaler()

X_train_mlp = scaler_X_mlp.fit_transform(X_train_full)
X_test_mlp = scaler_X_mlp.transform(X_test_full)
y_train_mlp = scaler_y_mlp.fit_transform(y_train_full.values.reshape(-1, 1)).flatten()

# 2. Initialize and train a Multi-Layer Perceptron (MLP)
mlp = MLPRegressor(
    hidden_layer_sizes=(64, 32),
    activation='relu',
    solver='adam',
    alpha=0.01,
    batch_size='auto',
    learning_rate='adaptive',
    max_iter=500,
    random_state=42,
    early_stopping=True,
    validation_fraction=0.1
)

mlp.fit(X_train_mlp, y_train_mlp)

# 3. Evaluate
y_pred_mlp_scaled = mlp.predict(X_test_mlp)
y_pred_mlp = scaler_y_mlp.inverse_transform(y_pred_mlp_scaled.reshape(-1, 1)).flatten()

r2_mlp = r2_score(y_test_full, y_pred_mlp)
mae_mlp = mean_absolute_error(y_test_full, y_pred_mlp)

print(f"--- Multi-Layer Perceptron (Neural Network) Results ---")
print(f"R-squared: {r2_mlp:.4f}")
print(f"MAE: {mae_mlp:.4f}")

# Safely handle potential NoneType for best_loss_
if hasattr(mlp, 'best_loss_') and mlp.best_loss_ is not None:
    print(f"Best loss reached: {mlp.best_loss_:.4f}")

print(f"\nImprovement over Stacking Ensemble: {r2_mlp - 0.1669:.4f}")
```
